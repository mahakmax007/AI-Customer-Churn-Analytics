```python
import pandas as pd
import numpy as np

df = pd.read_csv("Telco_Customer_Churn.csv")

print("Dataset Shape:", df.shape)
df.head()
```

```python
# Check column names
print("Columns:")
print(df.columns.tolist())

# Check data types and missing values
print("\nData Information:")
df.info()
```

```python
# Check missing values
print("Missing Values:")
print(df.isnull().sum())
```

```python
# Basic statistical summary
df.describe(include="all").T
```

```python
# Check the values that cannot be treated as numbers
total_charges_numeric = pd.to_numeric(df["TotalCharges"], errors="coerce")

print("Non-numeric TotalCharges values:",
      total_charges_numeric.isna().sum())

print("\nRows with non-numeric TotalCharges:")
print(df.loc[total_charges_numeric.isna(), ["customerID", "tenure", "MonthlyCharges", "TotalCharges", "Churn"]])
```

```python
# Check the churn distribution
print(df["Churn"].value_counts())

print("\nChurn percentage:")
print(df["Churn"].value_counts(normalize=True).mul(100).round(2))
```

```python
# Create a clean copy of the dataset
df_clean = df.copy()

# Convert TotalCharges from text to numeric
df_clean["TotalCharges"] = pd.to_numeric(
    df_clean["TotalCharges"],
    errors="coerce"
)

# Fill missing TotalCharges with 0
df_clean["TotalCharges"] = df_clean["TotalCharges"].fillna(0)

# Check the result
print("Missing TotalCharges after cleaning:",
      df_clean["TotalCharges"].isnull().sum())

print("TotalCharges data type:",
      df_clean["TotalCharges"].dtype)
```

```python
# Churn count
churn_count = df_clean["Churn"].value_counts()

print("Customer Churn Count:")
print(churn_count)

# Churn percentage
churn_percentage = (
    df_clean["Churn"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("\nCustomer Churn Percentage:")
print(churn_percentage)
```

```python
# Key churn KPIs
total_customers = len(df_clean)
churned_customers = (df_clean["Churn"] == "Yes").sum()
retained_customers = (df_clean["Churn"] == "No").sum()
churn_rate = (churned_customers / total_customers) * 100

print("Total Customers:", total_customers)
print("Churned Customers:", churned_customers)
print("Retained Customers:", retained_customers)
print("Churn Rate:", round(churn_rate, 2), "%")
```

```python
# Churn by contract type

contract_churn = pd.crosstab(
    df_clean["Contract"],
    df_clean["Churn"],
    normalize="index"
).mul(100).round(2)

print(contract_churn)
```

```python
# Create tenure groups

df_clean["TenureGroup"] = pd.cut(
    df_clean["tenure"],
    bins=[-1, 12, 24, 48, 72],
    labels=["0-12 Months", "13-24 Months", "25-48 Months", "49-72 Months"]
)

# Churn rate by tenure group

tenure_churn = pd.crosstab(
    df_clean["TenureGroup"],
    df_clean["Churn"],
    normalize="index"
).mul(100).round(2)

print(tenure_churn)
```

```python
# Customer counts by tenure group and churn

tenure_count = pd.crosstab(
    df_clean["TenureGroup"],
    df_clean["Churn"]
)

print(tenure_count)
```

```python
# Customer counts by Internet Service and Churn

internet_count = pd.crosstab(
    df_clean["InternetService"],
    df_clean["Churn"]
)

print(internet_count)
```

```python
# Create monthly charge groups

df_clean["MonthlyChargeGroup"] = pd.qcut(
    df_clean["MonthlyCharges"],
    q=4,
    labels=["Low", "Medium-Low", "Medium-High", "High"]
)

# Churn rate by monthly charge group

charge_churn = pd.crosstab(
    df_clean["MonthlyChargeGroup"],
    df_clean["Churn"],
    normalize="index"
).mul(100).round(2)

print(charge_churn)
```

```python
# Create an analytical customer risk score

df_clean["RiskScore"] = 0

df_clean["RiskScore"] += (
    df_clean["Contract"] == "Month-to-month"
).astype(int)

df_clean["RiskScore"] += (
    df_clean["tenure"] <= 12
).astype(int)

df_clean["RiskScore"] += (
    df_clean["InternetService"] == "Fiber optic"
).astype(int)

df_clean["RiskScore"] += (
    df_clean["PaymentMethod"] == "Electronic check"
).astype(int)

df_clean["RiskScore"] += (
    df_clean["OnlineSecurity"] == "No"
).astype(int)

df_clean["RiskScore"] += (
    df_clean["TechSupport"] == "No"
).astype(int)

# Convert score into risk category
df_clean["RiskCategory"] = pd.cut(
    df_clean["RiskScore"],
    bins=[-1, 1, 3, 6],
    labels=["Low Risk", "Medium Risk", "High Risk"]
)

print(df_clean[["customerID", "RiskScore", "RiskCategory", "Churn"]].head(10))
```

```python
import sqlite3

# Create a local SQLite database
conn = sqlite3.connect("customer_churn.db")

# Load the cleaned dataset into a SQL table
df_clean.to_sql(
    "customer_churn",
    conn,
    if_exists="replace",
    index=False
)

print("SQL database created successfully!")
print("Table: customer_churn")
```

```python
query = """
SELECT
    Churn,
    COUNT(*) AS customer_count,
    ROUND(COUNT(*) * 100.0 / (SELECT COUNT(*) FROM customer_churn), 2) AS percentage
FROM customer_churn
GROUP BY Churn;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT
    Contract,
    COUNT(*) AS total_customers,
    SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END) AS churned_customers,
    ROUND(
        SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS churn_rate
FROM customer_churn
GROUP BY Contract
ORDER BY churn_rate DESC;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT
    CASE
        WHEN tenure <= 12 THEN '0-12 Months'
        WHEN tenure <= 24 THEN '13-24 Months'
        WHEN tenure <= 48 THEN '25-48 Months'
        ELSE '49-72 Months'
    END AS tenure_group,
    COUNT(*) AS total_customers,
    SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END) AS churned_customers,
    ROUND(
        SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS churn_rate
FROM customer_churn
GROUP BY tenure_group
ORDER BY
    CASE tenure_group
        WHEN '0-12 Months' THEN 1
        WHEN '13-24 Months' THEN 2
        WHEN '25-48 Months' THEN 3
        WHEN '49-72 Months' THEN 4
    END;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT
    PaymentMethod,
    COUNT(*) AS total_customers,
    SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END) AS churned_customers,
    ROUND(
        SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS churn_rate
FROM customer_churn
GROUP BY PaymentMethod
ORDER BY churn_rate DESC;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT
    InternetService,
    COUNT(*) AS total_customers,
    SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END) AS churned_customers,
    ROUND(
        SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS churn_rate
FROM customer_churn
GROUP BY InternetService
ORDER BY churn_rate DESC;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT
    OnlineSecurity,
    TechSupport,
    COUNT(*) AS total_customers,
    SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END) AS churned_customers,
    ROUND(
        SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS churn_rate
FROM customer_churn
GROUP BY OnlineSecurity, TechSupport
ORDER BY churn_rate DESC;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
WITH risk_scored AS (
    SELECT
        CustomerID,
        Churn,
        (
            CASE WHEN Contract = 'Month-to-month' THEN 1 ELSE 0 END
            +
            CASE WHEN tenure <= 12 THEN 1 ELSE 0 END
            +
            CASE WHEN InternetService = 'Fiber optic' THEN 1 ELSE 0 END
            +
            CASE WHEN PaymentMethod = 'Electronic check' THEN 1 ELSE 0 END
            +
            CASE WHEN OnlineSecurity = 'No' THEN 1 ELSE 0 END
            +
            CASE WHEN TechSupport = 'No' THEN 1 ELSE 0 END
        ) AS risk_score
    FROM customer_churn
)

SELECT
    CASE
        WHEN risk_score <= 1 THEN 'Low Risk'
        WHEN risk_score <= 3 THEN 'Medium Risk'
        ELSE 'High Risk'
    END AS risk_category,
    COUNT(*) AS total_customers,
    SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END) AS churned_customers,
    ROUND(
        SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS churn_rate
FROM risk_scored
GROUP BY risk_category
ORDER BY
    CASE risk_category
        WHEN 'Low Risk' THEN 1
        WHEN 'Medium Risk' THEN 2
        WHEN 'High Risk' THEN 3
    END;
"""

result = pd.read_sql_query(query, conn)
result
```
